```python
def train_linear_regression(X, y):
    if not (X[:,0] == np.ones(X.shape[0])).sum() == X.shape[0]:
        X = np.column_stack([np.ones(X.shape[0]), X])

    XTX = X.T.dot(X)
    XTX_inv = np.linalg.inv(XTX)
    w = XTX_inv.dot(X.T).dot(y)

    return w

# Example
X = df[["engine_hp", "highway_mpg", "popularity"]]
X = np.array(X)

y = df[["msrp"]]
y = np.array(y)

w = train_linear_regression(X, y)
w
```

# RMSE

```python
def rmse(y, y_pred):
   
    error = np.sqrt(((y - y_pred)**2).mean())
    return error
```

```python
df.columns

base = ["engine_hp", "engine_cylinders", "highway_mpg", "city_mpg", "popularity"]

def prepar_X(df):
    df_num = df[base]
    df_num = df_num.fillna(0)
    X = df_num.values
    X = np.column_stack([np.ones(X.shape[0]), X])
    return X

X_train = prepar_X(df_train)
X_train 
```
